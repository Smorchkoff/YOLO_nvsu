## Задание 19

#### Ответ: -47.5047

```python
X = [428, 300, 250, 578, 492, 204, 245, 420, 760, 363, 600, 483,
     711, 265, 572, 350, 652, 460, 340, 670, 565, 750, 452]
Y = [21, 33, 43, 71, 85, 17, 38, 43, 53, 76, 33, 16, 24, 59, 65,73, 45, 63, 53, 43, 58, 28, 37]
n = len(X)

# 1. Средние
mean_x = sum(X) / n
mean_y = sum(Y) / n

# 2. Сумма произведений отклонений
sum_product = 0
for i in range(n):
    sum_product += (X[i] - mean_x) * (Y[i] - mean_y)

# 3. Ковариация (деление на n)
cov_xy = sum_product / n - 1

# 4. Округление до 4 знаков
result = round(cov_xy, 4)

print("Ковариация cov(X, Y) =", result)
```

## Задание 20

#### Ответ: -0.0146

```python

# Суммы для ковариации и дисперсий
sum_cov = 0
sum_var_x = 0
sum_var_y = 0

for i in range(n):
    dx = X[i] - mean_x
    dy = Y[i] - mean_y
    sum_cov += dx * dy
    sum_var_x += dx * dx
    sum_var_y += dy * dy

# Ковариация и стандартные отклонения (деление на n)
cov_xy = sum_cov / n
std_x = (sum_var_x / n) ** 0.5
std_y = (sum_var_y / n) ** 0.5

# Коэффициент Пирсона
pearson_r = cov_xy / (std_x * std_y)

# Округление до 4 знаков
result = round(pearson_r, 4)
print("Коэффициент Пирсона =", result)
```
