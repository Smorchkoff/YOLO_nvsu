## 1 Задание
### Реализуйте функцию inverse(lst), которая “переворачивает” список

```python
def inverse(lst):
    return lst[::-1]
print(inverse('Hello'))
print(inverse([1, 2, 3]))
```

## 2 Задание
### Реализуйте функцию tokenize(str), которая принимает на вход строку, а возвращает список слов, которые содержатся в этой строке. Известно, что слова разделены одним или несколькими пробелами

```python
def tokenize(str) -> list:
    return str.split()
print(tokenize('One Two Three Four'))
print(tokenize('Два  пробела  между  словами'))
```

## 3. Задание
### Дан список, состоящий из кортежей вида
### ('Maxim', 1996)
### где на первом месте стоит имя, на втором - дата рождения. С использованием спискового включения (listcomprehention), реализуйте функцию, которая принимает на вход список, и возвращает измененный и отфильтрованный список:
### 1. Имя должно быть перевернуто
### 2. Вместо года рождение нужно оставить число - сколько лет исполнилось человеку в 2010 году (может быть отрицательным)
### 3. В списке должны быть оставлены только те люди, у кого год рождения - четное число


```python
people = [('Maxim', 1996), ('Anton', 2002), ('OLEG', 1993)]

def reverse_people(ls: list) -> list:
    return [(name[::-1], 2010-year) for name, year in ls if year % 2 == 0]
print(reverse_people(people))
```

## 4. Задание
### Реализуйте функцию word_count(s), которая подсчитывает сколько раз какое слово встретилось в тексте. Слова разделены одним пробелом. Функция должна возвращать словарь, ключи которого - слова, значения - число


```python
def word_counts(s: str) -> dict:
    words = s.split()
    word_count = {}
    for word in words:
        word_count[word] = words.count(word)
    return word_count
print(word_counts('aa bb cc aa bb aa'))
```
