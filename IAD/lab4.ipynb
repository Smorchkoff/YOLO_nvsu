### Евклидово расстояние

```python
from math import sqrt

coords = [[1, 1, 1, 1, 10],
          [2, 2, 2, 4, 5],
          [10, -15, -11, 9, 9],
          [0.06297183, 0.77261217, 0, 0.94396506, 0.107809358],
          [41, 58, 72, 72, 35]]

def euclid(object1, object2):
    sum = 0
    for x1,x2 in zip(object1, object2):
        sum += (x1-x2)**2
    sum = sqrt(sum)
    return sum

for i1,coord1 in enumerate(coords):
    for i2,coord2 in enumerate(coords):
        print(f'Евклидово расстояние от объекта {i1} до объекта {i2}: {euclid(coord1, coord2)}')


```

### Квадрат евклидова расстояния

```python
def euclid_qrt(object1, object2):
    sum = 0
    for x1,x2 in zip(object1, object2):
        sum += (x1-x2)**2
    return sum
for i1,coord1 in enumerate(coords):
    for i2,coord2 in enumerate(coords):
        print(f'Квадрат евклидова расстояния от объекта {i1} до объекта {i2}: {euclid_qrt(coord1, coord2)}')
```

### Манхэттенское расстояние

```python
def manhattan(object1, object2):
    sum = 0
    for x1,x2 in zip(object1, object2):
        sum += abs(x1-x2)
    return sum

for i1,coord1 in enumerate(coords):
    for i2,coord2 in enumerate(coords):
        print(f'Манхэттенское расстояние от объекта {i1} до объекта {i2}: {euclid_qrt(coord1, coord2)}')
```

### Расстояние Чебышева

```python
def chebishev(object1, object2):
    sum = []
    for x1,x2 in zip(object1, object2):
        sum.append(abs(x1-x2))
    return max(sum)

for i1,coord1 in enumerate(coords):
    for i2,coord2 in enumerate(coords):
        print(f'Расстояние Чебышева от объекта {i1} до объекта {i2}: {euclid_qrt(coord1, coord2)}')
```
